# Notebook 04: Comparative Model Evaluation & Research Question Experiments

This notebook analyzes the experimental results addressing all 6 FOML research questions: traditional ML vs CNN, underwater optical degradation, class imbalance, and cross-domain transfer.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import json
import pandas as pd
import matplotlib.pyplot as plt

with open('../outputs/metrics/test_metrics_mavecodd.json', 'r') as f:
    metrics = json.load(f)

print('Evaluation summary loaded successfully.')
```

## 1. Traditional ML Baseline vs Deep Learning U-Net


```python
base = metrics['baseline_model_a']
unet = metrics['deep_learning_model_b']

df = pd.DataFrame([
    {'Metric': 'Accuracy', 'Baseline SVM': base['accuracy'], 'U-Net': unet['classification_accuracy']},
    {'Metric': 'Precision', 'Baseline SVM': base['precision'], 'U-Net': unet['classification_precision']},
    {'Metric': 'Recall', 'Baseline SVM': base['recall'], 'U-Net': unet['classification_recall']},
    {'Metric': 'F1-Score', 'Baseline SVM': base['f1_score'], 'U-Net': unet['classification_f1']},
])
print(df.to_string(index=False))
```

## 2. Optical Turbidity Degradation Stress Test


```python
deg = pd.DataFrame(metrics['optical_degradation_stress_test'])
plt.figure(figsize=(8, 4.5))
plt.plot(deg['turbidity'], deg['mean_iou'], marker='o', label='Mean IoU')
plt.plot(deg['turbidity'], deg['mean_dice'], marker='s', label='Mean Dice')
plt.xlabel('Turbidity Scattering Coefficient')
plt.ylabel('Score')
plt.title('Performance vs Underwater Turbidity')
plt.legend()
plt.grid(True)
plt.show()
```
