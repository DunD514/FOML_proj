# Notebook 02: Traditional ML Baseline (HOG + Color + PCA + SVM)

In this notebook, we train and evaluate **Model A**, our handcrafted feature baseline. We extract Histogram of Oriented Gradients (HOG) features and HSV color histograms, reduce dimensions with PCA, and fit an SVM.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from src.utils.config import load_config
from src.data.split_dataset import load_split_manifest
from src.models.baseline import HullDefectBaselineClassifier, prepare_patch_dataset
from src.models.evaluate import evaluate_baseline_on_patches

config = load_config()
splits = load_split_manifest('../data/processed/split_manifest_mavecodd.json')
print(f'Train scenes: {len(splits["train"])}, Test scenes: {len(splits["test"])}')
```

## 1. Extract Patches and Handcrafted Features


```python
train_imgs, train_y = prepare_patch_dataset(splits['train'], patches_per_sample=4)
print(f'Total training patches: {len(train_imgs)}')

model = HullDefectBaselineClassifier(pca_components=64, svm_c=10.0)
X_train = model.extract_batch_features(train_imgs)
print('Feature matrix shape:', X_train.shape)
model.fit(X_train, train_y)
print('Baseline model trained successfully!')
```

## 2. Evaluate Baseline on Held-out Test Set


```python
eval_res = evaluate_baseline_on_patches(model, splits['test'])
for k, v in eval_res.items():
    print(f'{k}: {v}')
```
